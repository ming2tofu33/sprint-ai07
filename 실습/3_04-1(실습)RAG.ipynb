```python
MODEL_ID = 'google/gemma-3-1b-it'
```

```python
pipe = HuggingFacePipeline.from_model_id(
    model_id = MODEL_ID,
    task = 'text-generation',
    pipeline_kwargs= {
        'max_new_tokens' : 256,
        'do_sample' : True,
        'temperature' : 0.7,
        'return_full_text' :False
    }
)
```

```python
llm = ChatHuggingFace(llm=pipe)
```

```python
def ask_expert(persona, question):
    messages = [SystemMessage(content = f"당신은 {persona}입니다"), 
            HumanMessage(content = question)]

    response = llm.invoke(messages)
    return {'persona' : persona, 'answer' : response.content.strip()
           }
```

```python
from datetime import datetime
def add_context(input_dict):
    input_dict['today'] = datetime.now().strftime('%Y년 %m월 %d일')
    return input_dict
    # {'question' : '오늘 날짜에 맞는 인사말을 해줘', 'today' : '2026년 3월 25일'}

prompt = ChatPromptTemplate.from_messages([
    ('system', '오늘 날짜는 {today}입니다'),
    ('human', '{question}')
])

chain = (
    RunnableLambda(add_context) | prompt | llm | StrOutputParser())

chain.invoke({'question' : '오늘 날짜에 맞는 인사말을 해줘'})

# chain2 = prompt | llm | StrOutputParser()
# chain2.invoke({'question' : '오늘 날짜에 맞는 인사말을 해줘', 'today' : '2026년 3월 25일'})
```

```python
positive_chain = (
    ChatPromptTemplate.from_messages([
        ('human', '{topic}의 장점 3가지를 간단히 나열해줘')
    ])
    | llm 
    | StrOutputParser()
)

negative_chain = (
    ChatPromptTemplate.from_messages([
        ('human', '{topic}의 단점 3가지를 간단히 나열해줘')
    ])
    | llm 
    | StrOutputParser()
)

parallel_chain = RunnableParallel(pros = positive_chain, cons =negative_chain)
```

```python
def search_documents(document_list, key, value) :
    return [doc for doc in document_list if doc.metadata.get(key) == value]

def get_doc_stats(document_list):
    all_keys = set()
    for doc in document_list:
        all_keys.update(doc.metadata.keys())
    lengths = [len(doc.page_content) for doc in document_list]
    return {
        'total' : len(document_list),
        'avg_length' : sum(lengths) / len(lengths) if lengths else 0,
        'metadata_keys' : sorted(all_keys)
    }
    # 전체 문서의 수, 평균 길이, 모든 metadata 키 목록을 리턴
```
