```python
# z : [1, 2, 3, 4, 5]
# T : 1,000,000

# softmax(1/1M, 2/1M, ... 5/1M)   : distribution이 flat : 좀더 다양한 값이 출력된다               -> 다양한 답변이 생성된다.

# T : 0.1
# softmax(10, 20, 30, 40, 50)     : distribution이       : 큰값만 계속 뽑힌다

# max_p : 0.9
# 상위 30%의 토큰들만 가지고 weighted sampling을 합니다.
# 2, 3, 4, 5 :
# 2 * 1, 3 * 1.01, 4 * 1.011, 5 * 1.02  (
# 2 * 1000 , 3 * 1010 , 4 * 1011, 5 * 1020

# GPT : transformer decoder :
# 40, 50 : 40 * 1, 50 * 8
```

```python
def call_llm(prompt, system="", temperature=1.0, max_tokens=1024, model='gpt-4o-mini'):
    messages = []
    if system:
        messages.append({'role':'system', 'content':system})
    messages.append({'role':'user', 'content': prompt})
    response = client.chat.completions.create(
        model = model,
        messages = messages,
        max_tokens = max_tokens,
        temperature = temperature
    )

    return response.choices[0].message.content
```

```python
import time
import numpy as np

chunks = []
times =  []

start =  time.time()

stream = client.chat.completions.create(
    model = 'gpt-4o-mini',
    messages = [{"role" : "user", "content" : '1부터 10까지 세어주세요'}],
    stream=True
)

text = ''  # int : 4byte, char : 4byte, float : 8byte , fp16, fp8
for chunk in stream:
    if chunk.choices[0].delta.content:
        text = chunk.choices[0].delta.content   # 초기화 + 변수 할당   -> text 라는 공간을 만들어서 -> 그 공간에 chunk.choices[0].delta.content 값을 할당
        print(text, end='', flush=True)

        chunks.append(text)
        times.append(time.time() - start)  # 10, 11, 12, ... -> 10, 1, 1, 1,


intervals = [times[i] - times[i-1] for i in range(1, len(times))]

print("------------------")
print(len(chunks), np.mean(intervals), np.min(intervals), np.max(intervals))
```

```python
headlines = [
    "핵심 반도체 설계 도면 외부 유출 정황 포착",
    "내일 오전 사내 카페테리아 정기 소독 안내",
    "2분기 영업이익 전년 대비 15% 상승 발표",
    "신규 프로젝트 '제네시스' 팀원 명단 공개",
    "사내 주차장 차단기 교체 공사 진행"
]

# 보안등급 : Lv1(일반), Lv2(대외비), Lv3(기밀)
# 부서코드 : Sec(보안), HR(인사), PR(홍부), FAC(시설), RD(연구소)

few_shot_examples = """
    다음 뉴스 헤드라인을 사내 보안등급과 부서코드에 맞게 분류해주세요.
    형식 : [보안등급/부서코드]
    - 보안등급 : Lv1(일반), Lv2(대외비), Lv3(기밀)
    - 부서코드 : Sec(보안), HR(인사), PR(홍보), FAC(시설), RD(연구소)

    예시 :
    헤드라인 : "신제품 출시 기념 전사 휴무 검토"
    결과 : [Lv1/HR]

    헤드라인 : "차세대 배터리 실험 데이터 백업 지침"
    결과 : [Lv3/RD]

    헤드라인 : "언론 보도용 자료 초안 작성"
    결과 : [Lv2/PR]
    """

for headline in headlines:
    prompt = (f"{few_shot_examples}\n"
             f"헤드라인: \"{headline}\"\n"
            "결과: "
    )
    response = client.chat.completions.create(
        model = 'gpt-4o-mini',
        messages = [{"role" : "user", "content" : prompt}],
    )

    print(response.choices[0].message.content)
```

```python
REVIEW_TEMPLATE = """다음 {product} 리뷰를 분석하세요.

<review>
{review_text}
</review>

분석 항목 : {aspects}
아래 json 형식으로 출력하세요:
{{
    "overall_sentiment" : "긍정/부정/중립",
    "aspects" : {aspect_schema},
    "summary" : "한 줄 요약"
}}

JSON만 출력하세요.
"""
```

```python
aspects = ['컴퓨터', '핸드폰', 'TV']
', '.join(aspects)
```

```python
def review_template(product, review_text, aspects):
    aspects_str = ", ".join(aspects)
    schema =  "{" + ", ".join([f'"{a}": "긍정/부정/중립"' for a in aspects]) + "}"
    return REVIEW_TEMPLATE.format(
        product = product,
        review_text = review_text,
        aspects = aspects_str,
        aspect_schema = schema
    )
```

```python
review_template('노트북', '화면이 선명하고 배터리도 오래갑니다', ['화면', '키보드', '배터리'])
```

```python
class PromptTemplate:
    def __init__(self, name, system, template, defaults=None):
        self.name = name
        self.system = system
        self.template = template
        self.defaults = defaults

    def render(self, **kwargs):
        params = {**self.defaults, **kwargs}
        return self.template.format(**params)

    def run(self, max_tokens = 300, **kwargs):
        print(kwargs)
        prompt = self.render(**kwargs)
        return client.chat.completions.create(
                model = 'gpt-4o-mini',
                max_tokens=max_tokens,
                messages = [
                    {"role" : "system", "content" : self.system},
                    {"role" : "user", "content" : prompt}
                ]
            )
```

```python
summarizer = PromptTemplate(
    name = "텍스트 요약가",
    system = "당신은 전문 요약 작가입니다. 핵심만 간결하게 요약합니다.",
    template = "다음 텍스트를 {num_sentences}문장으로 요약하세요.\n\n<text>\n{text}\n</text>\n\n{num_sentences}문장 요약:",
    defaults = {"num_sentences":3}
)
```

```python
# self - consistency
# ensemble

# 같은 질문 -> 모델 5번 -> 다수결 (답변의 다양성 높아야 : temperature 높여서)
```

```python
from collections import Counter
import re

def self_consistency(prompt, n=5):
    answers = []
    full_responses = []

    extract_prompt = prompt + "\n\n 단계별로 풀어주세요. 마지막 줄에 '최종 답: [답]' 형식으로 작성해주세요"

    for i in range(n):
        response = client.chat.completions.create(
            model = 'gpt-4o-mini',
            max_tokens=800,
            messages = [
                {"role" : "user", "content" : extract_prompt}
            ]
        ).choices[0].message.content
        full_responses.append(response)
        match = re.search(r"최종 답[:\s]+(.+)", response)
        print("match", match)
        if match:
            answers.append(match.group(1).strip())
        else:
            answers.append(response.strip().split("\n")[-1])

    counter = Counter(answers)
    best_answer, best_count = counter.most_common(1)[0]
    confidence = best_count / n

    return {
        "answer" : best_answer,
        "confidence" : confidence,
        "votes" : dict(counter),
        "responses" : full_responses
    }
```

```python
# 다음 예시문을 보고 한글로 번역하고, 거기서 2줄로 요약하고, 거기서 3개 키워드를 뽑으세요
# <text> ~~~ </text>

# step1 : 기사를 3줄로 요약 -> 요약문으로
# step2 : 요약문의 감성을 분석해서 -> 감정
# step3 : 감정과 연관된 핵심 키워드 추출 -> 키워드

# sequential : step1 -> step2 -> stepN...
# branch : step1 ---------step2_a
#                 |-------step2_b
# parallel : step1 -------(step2_a, step2_b, step2_c) ------ step3

# refine :  step1 - step2 - 검증 (실패) - step1
```

```python
article = """삼성전자가 3나노 공정 기반의 차세대 AI 반도체를 공개했다.
        이번 칩은 기존 대비 전력 소모를 40% 줄이면서 성능은 2배 향상됐다.
        업계에서는 엔비디아의 독주에 제동을 걸 수 있을지 주목하고 있다.
        다만 양산 일정이 불확실하다는 점은 리스크로 지적된다."""

step1 = client.chat.completions.create(
    model = 'gpt-4o-mini',
    max_tokens=800,
    messages = [
        {"role" : "user", "content" : f"다음 기사를 한 문장으로 요약해주세요:\n\n{article}"}
    ]
).choices[0].message.content

step2 = client.chat.completions.create(
    model = 'gpt-4o-mini',
    max_tokens=800,
    messages = [
        {"role" : "user", "content" : f"다음 문장의 전반적인 논조를 '긍정', '부정', '중립' 중 하나로 판단하고 이유를 설명해주세요:\n\n{step1}"}
    ]
).choices[0].message.content

step3 = client.chat.completions.create(
    model = 'gpt-4o-mini',
    max_tokens=800,
    messages = [
        {"role" : "user", "content" : f"다음 기사에서 핵심 키워드 5개를 쉼표로 구분해서 추출해주세요:\n\n{article}\n\n다음 요약을 이용하세요\n{step1}"}
    ]
).choices[0].message.content

step4 = client.chat.completions.create(
    model = 'gpt-4o-mini',
    max_tokens=800,
    messages = [
        {"role" : "user", "content" : f"다음 분석 결과를 바탕으로 한 문단의 브리핑 리포트를 작성해주세요:\n- 요약 : {step1}\n- 감정 : {step2}\n- 키워드 : {step3}"}
    ]
).choices[0].message.content
```
