# 1. 최종 제출물 이해하기

이 미션은 단순히 모델을 학습하는 과제가 아니라, 두 명의 연구자가 Docker를 사용해서 작업을 넘겨주는 흐름을 보여주는 과제입니다.

최종 제출물은 크게 두 가지입니다.

- `mission-result` 코드 폴더
- 2페이지 이내 보고서 PDF

`mission-result` 폴더에는 연구자 1의 학습 관련 파일, 연구자 2의 추론 관련 파일, Dockerfile, `docker-compose` 설정, 결과 파일이 들어가야 합니다.

보고서에는 반드시 다음 내용이 들어가야 합니다.

- Docker Hub URL
- 연구자 1의 데이터 전처리 및 모델링 결과 요약
- 코드 아키텍처 도식 및 설명

따라서 노트북에서 모델 성능만 잘 내는 것으로 끝내면 안 되고, 파일이 Docker 컨테이너 사이에서 어떻게 전달되는지도 설명할 수 있어야 합니다.


# 2. Docker 초보자를 위한 핵심 개념 정리

이번 미션에서 Docker를 이해할 때는 아래 개념만 먼저 잡으면 됩니다.

| 개념 | 이번 미션에서의 의미 |
| --- | --- |
| Dockerfile | 어떤 Python 환경을 만들고, 어떤 파일을 넣고, 무엇을 실행할지 적는 설명서 |
| image | Dockerfile로 만든 실행 환경 묶음 |
| container | image를 실제로 실행한 상태 |
| Docker Hub | 내가 만든 image를 업로드하고 다른 사람이 내려받을 수 있는 저장소 |
| volume | 컨테이너와 호스트 또는 컨테이너끼리 파일을 공유하는 통로 |
| docker-compose | 여러 컨테이너를 한 번에 실행하고 연결하는 설정 파일 |

이번 과제에서 연구자 1 컨테이너는 `train.csv`로 학습해서 `model.pkl`을 만드는 역할입니다.

연구자 2 컨테이너는 연구자 1이 만든 `model.pkl`과 `test.csv`를 받아서 예측 결과인 `result.csv`를 만드는 역할입니다.

중요한 포인트는 연구자 2가 처음부터 데이터와 모델을 가지고 있으면 안 된다는 점입니다. 연구자 2는 연구자 1의 Docker Hub 이미지 또는 컨테이너 실행 결과를 통해 필요한 파일을 받아야 합니다.


# 3. 작업 폴더 구조 설계하기

코드를 작성하기 전에 제출 폴더 구조를 먼저 정해야 합니다. 폴더 구조가 명확해야 Dockerfile, 데이터, 모델 파일, 추론 결과가 섞이지 않습니다.

추천 구조는 다음과 같습니다.

- `mission-result/researcher1`: 연구자 1의 학습 스크립트, Dockerfile, requirements 파일
- `mission-result/researcher2`: 연구자 2의 `inference.ipynb`, 추론 관련 파일
- `mission-result/shared`: 컨테이너끼리 공유할 `model.pkl`, `test.csv`, `result.csv` 위치
- `mission-result/docker-compose.yml`: 연구자 1 컨테이너와 Jupyter Notebook 컨테이너를 함께 실행하는 설정

이 구조를 쓰면 역할이 분명해집니다.

- 연구자 1 영역: 학습과 모델 저장
- 연구자 2 영역: 모델 로드와 추론
- shared 영역: 두 연구자 사이의 파일 전달

보고서의 아키텍처 도식도 이 폴더 구조를 기준으로 그리면 설명하기 쉽습니다.

# 5. 연구자 1: EDA 계획 세우기

EDA는 모델링 전에 데이터를 이해하는 단계입니다. 보고서에는 긴 분석을 모두 넣을 필요는 없지만, 어떤 데이터 특성이 있었고 어떤 전처리를 했는지는 설명할 수 있어야 합니다.

이 단계에서 확인할 내용은 다음과 같습니다.

- 전체 행과 열 개수
- 각 컬럼의 데이터 타입
- 결측치 개수
- 숫자형 변수의 평균, 표준편차, 최소값, 최대값
- `Extracurricular Activities` 값의 종류와 개수
- 목표변수 `Performance Index`의 분포
- 입력 변수와 목표변수의 관계

특히 이 데이터에서는 `Previous Scores`, `Hours Studied`가 `Performance Index`와 강한 관련이 있을 가능성이 큽니다. 이런 관계를 확인하면, 왜 회귀 모델을 사용했는지 보고서에서 설명하기 좋습니다.

EDA 결과는 너무 길게 정리하지 말고, 나중에 보고서에 넣을 수 있도록 핵심 문장 3개 정도로 요약하는 것을 목표로 합니다.


# 6. 연구자 1: 전처리와 모델링 계획 세우기

이 미션의 목표는 회귀 모델로 `Performance Index`를 예측하는 것입니다.

전처리에서 가장 중요한 부분은 범주형 변수 처리입니다.

- 숫자형 변수: 그대로 사용 가능
- `Extracurricular Activities`: Yes/No 형태라면 숫자로 바꾸거나 원-핫 인코딩 필요

모델링 단계에서는 처음부터 복잡한 모델을 많이 쓰기보다, 설명하기 쉬운 모델부터 비교하는 것이 좋습니다.

추천 비교 대상은 다음 정도입니다.

- 선형 회귀 계열 모델
- 규제가 들어간 선형 모델
- 트리 기반 회귀 모델

평가지표는 미션에서 요구한 RMSE를 사용합니다. RMSE는 예측값과 실제값의 차이를 원래 점수 단위로 해석할 수 있어서, `평균적으로 몇 점 정도 오차가 났는지` 설명하기 좋습니다.

최종 모델은 RMSE가 낮고, Docker 환경에서 안정적으로 저장하고 불러올 수 있는 모델을 선택하면 됩니다.


# 7. 연구자 1: `model.pkl` 저장 전략 정하기

`model.pkl`에는 단순히 학습된 모델만 저장하는 것보다, 전처리 과정까지 함께 저장하는 것이 안전합니다.

이유는 연구자 2가 추론할 때도 연구자 1과 같은 전처리를 해야 하기 때문입니다. 전처리 방식이 조금이라도 달라지면 예측 결과가 달라지거나 오류가 날 수 있습니다.

따라서 추천 전략은 다음과 같습니다.

- 입력 컬럼 목록을 명확히 정한다.
- 범주형 컬럼 처리 방식을 모델 저장 객체 안에 포함한다.
- 전처리와 모델을 하나의 파이프라인처럼 묶어서 저장한다.
- 저장된 `model.pkl`을 다시 불러와서 예측이 되는지 반드시 확인한다.

보고서에는 `전처리와 모델을 함께 저장하여 연구자 2의 추론 환경에서도 동일한 입력 처리 과정을 재현하도록 설계했다`고 설명하면 좋습니다.


# 8. 연구자 1: 학습 과정을 `.py` 파일로 정리하기

Notebook은 실험과 설명에는 좋지만, Docker 컨테이너에서 자동 실행하기에는 `.py` 스크립트가 더 적합합니다.

따라서 이 Notebook에서 검증한 흐름을 나중에 하나의 학습 스크립트로 정리해야 합니다.

학습 스크립트가 담당할 일은 다음과 같습니다.

- `train.csv` 읽기
- 입력 변수와 목표변수 나누기
- 전처리 방식 정의
- 학습용과 검증용 데이터 분리
- 모델 학습
- RMSE 계산 및 출력
- 전체 학습 데이터로 최종 모델 재학습
- `model.pkl` 저장
- 연구자 2가 사용할 수 있도록 `test.csv`와 `model.pkl`을 공유 위치에 준비

이 스크립트는 연구자 1 Docker 컨테이너가 실행될 때 자동으로 수행될 작업의 중심이 됩니다.


# 9. 연구자 1: Docker 이미지 만들기

연구자 1의 Docker 이미지는 `학습을 자동으로 수행하는 실행 환경`입니다.

이 이미지 안에는 다음이 포함되어야 합니다.

- Python 실행 환경
- 필요한 패키지 목록
- 학습 스크립트
- `train.csv`
- `test.csv`
- 컨테이너 실행 시 학습 스크립트를 실행하는 설정

이 단계에서 중요한 것은 컨테이너를 실행했을 때 `model.pkl`이 실제로 생성되는지 확인하는 것입니다.

초보자 기준으로는 Docker 이미지를 한 번에 완성하려고 하기보다, 다음 순서로 확인하는 것이 좋습니다.

1. 로컬 Python 환경에서 학습 스크립트가 정상 작동하는지 확인
2. Docker 이미지가 정상적으로 만들어지는지 확인
3. 컨테이너 실행 시 학습 스크립트가 자동 실행되는지 확인
4. 컨테이너 안 또는 공유 폴더에 `model.pkl`이 생기는지 확인

이 과정을 통과하면 연구자 1의 역할은 대부분 완성된 것입니다.


# 10. 연구자 1: Docker Hub에 이미지 업로드하기

연구자 1의 이미지는 Docker Hub에 업로드해야 합니다. 그래야 연구자 2가 사전에 파일을 가지고 있지 않아도, Docker Hub 이미지를 통해 필요한 환경과 파일을 가져올 수 있습니다.

이 단계에서 해야 할 일은 다음과 같습니다.

- Docker Hub 계정 준비
- 연구자 1 이미지 이름 정하기
- 로컬에서 이미지가 정상 실행되는지 최종 확인
- Docker Hub에 이미지 업로드
- 업로드된 이미지 URL 기록

보고서에는 Docker Hub URL을 반드시 적어야 합니다.

이미지 이름은 과제 식별이 쉬운 이름으로 정하는 것이 좋습니다. 예를 들어 `mission15-researcher1`처럼 역할과 미션 번호가 드러나는 이름이 좋습니다.

주의할 점은 Docker Hub에 업로드한 이미지가 실제로 다른 환경에서도 내려받아 실행 가능한 상태여야 한다는 것입니다.


# 11. 연구자 2: `docker-compose`로 협업 환경 구성하기

연구자 2는 연구자 1이 만든 Docker Hub 이미지와 별도의 Jupyter Notebook 이미지를 함께 사용해야 합니다.

`docker-compose`의 역할은 두 컨테이너를 하나의 협업 환경처럼 묶는 것입니다.

구성해야 할 컨테이너는 다음과 같습니다.

- 연구자 1 컨테이너: Docker Hub 이미지 기반, `model.pkl`과 `test.csv` 제공
- 연구자 2 Jupyter 컨테이너: `inference.ipynb` 실행, `result.csv` 생성

파일 전달 전략은 두 가지를 생각할 수 있습니다.

첫 번째는 공유 볼륨 전략입니다. 연구자 1 컨테이너와 연구자 2 컨테이너가 같은 `shared` 위치를 바라보게 해서, 연구자 1이 만든 `model.pkl`을 연구자 2가 바로 읽을 수 있게 합니다.

두 번째는 `docker cp` 전략입니다. 연구자 1 컨테이너 내부에서 생성된 `model.pkl`과 `test.csv`를 호스트의 `shared` 폴더로 복사한 뒤, 연구자 2 Jupyter 컨테이너가 그 폴더를 읽도록 합니다.

미션 힌트에 두 방식이 모두 언급되어 있으므로, 실제 구현에서는 둘 중 하나를 중심 전략으로 정하고 보고서에서 명확히 설명하면 됩니다.


# 12. 연구자 2: `inference.ipynb`에서 추론하기

`inference.ipynb`는 연구자 2의 핵심 산출물입니다.

이 Notebook에서는 학습을 다시 하면 안 됩니다. 연구자 2의 역할은 이미 만들어진 `model.pkl`을 받아서 `test.csv`에 대한 예측을 수행하는 것입니다.

`inference.ipynb`에 들어갈 흐름은 다음과 같습니다.

- 공유 위치에서 `model.pkl`이 존재하는지 확인
- 공유 위치에서 `test.csv`가 존재하는지 확인
- `model.pkl` 로드
- `test.csv` 로드
- 예측 수행
- 예측 결과를 `result.csv`로 저장
- 저장된 `result.csv`의 행 개수가 `test.csv`와 같은지 확인

결과 파일에는 최소한 예측값 컬럼이 있어야 합니다. 가능하면 행 순서가 원본 `test.csv`와 동일하게 유지되도록 주의합니다.

이 단계가 끝나면 연구자 2의 역할은 완성됩니다.
