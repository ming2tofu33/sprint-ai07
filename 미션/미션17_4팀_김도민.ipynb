# 미션 소개
이번 미션에서는 사용자가 웹 인터페이스에서 손으로 숫자를 그려 입력하면 해당 숫자를 예측하는 Streamlit 기반 웹 서비스를 개발해봅시다. 이를 위해 [Github의 ONNX 모델 저장소](https://github.com/onnx/models/tree/main/validated/vision/classification/mnist/model)에서 제공하는 MNIST ONNX 모델을 활용해볼거에요.
이번 미션을 해결하면, 여러분들은 모델을 구현하는 것에 끝나지않고 직접 AI 기반 서비스를 만드는 엔지니어로 성장할 수 있어요.

# 가이드라인
1. 사용자 인터페이스: 서비스는 다음과 같은 4개의 주요 화면 영역으로 구성됩니다.
   - 입력 캔버스: 사용자가 마우스로 숫자를 그릴 수 있는 캔버스 영역(streamlit-drawable-canvas 라이브러리 활용)
   - 전처리 이미지 표시: 사용자가 그린 숫자 이미지를 모델 추론에 적합하게 전처리한 결과를 시각적으로 표시
   - 모델 추론 결과: 0부터 9까지의 각 레이블에 대한 예측 확률을 막대 차트(bar chart) 등으로 시각화
   - 이미지 저장소: 사용자가 그린 이미지를 저장하고, 각 이미지와 함께 예측된 레이블 및 확률을 표시
2. 필수 기능:
   - 모델 관리: MNIST ONNX 모델을 다운로드하고, 세션 간 효율적인 모델 관리를 위해 캐싱 기능 구현
   - 이미지 처리 및 추론: 사용자가 그린 이미지를 모델 입력 사양에 맞게 전처리하고, 모델 추론을 수행하는 함수 개발
3. 배포:
   - 도커 컨테이너화: Dockerfile을 작성하여 서비스를 도커 이미지로 빌드
   - 도커 허브 배포: 완성된 도커 이미지를 Docker Hub에 업로드하여 배포

완료 기준은 다음과 같습니다.

- 이번 미션이 노트북 중심이 아니라 Streamlit 서비스 중심 과제라는 점을 이해했다.


# 2. 이번 미션에서 공부할 핵심 개념 정리하기

이번 미션은 AI 모델을 학습하는 방법보다 `모델을 서비스로 만드는 방법`을 공부하기 위한 과제입니다.

가장 중요한 개념은 네 가지입니다.

첫째, ONNX 모델 사용입니다. ONNX는 PyTorch나 TensorFlow 같은 특정 프레임워크에 묶이지 않고 모델을 실행할 수 있게 해주는 모델 형식입니다. 이번 미션에서는 MNIST ONNX 모델을 불러와 추론에 사용합니다.

둘째, 이미지 전처리입니다. 사용자가 캔버스에 그린 이미지는 모델이 바로 이해할 수 있는 형태가 아닙니다. 모델이 요구하는 28x28 크기, 흑백 이미지, 정규화된 숫자 배열 형태로 바꾸어야 합니다.

셋째, Streamlit 웹 인터페이스입니다. 사용자가 직접 숫자를 그리고, 전처리 결과와 예측 확률을 눈으로 확인할 수 있게 화면을 구성해야 합니다.

넷째, Docker 배포입니다. 완성한 서비스를 내 컴퓨터에서만 실행하는 것이 아니라 Docker 이미지로 만들어 배포하는 과정을 경험합니다.

완료 기준은 다음과 같습니다.

- ONNX가 무엇을 위한 형식인지 설명할 수 있다.
- 캔버스 이미지가 전처리를 거쳐야 하는 이유를 설명할 수 있다.
- Streamlit이 이번 미션에서 어떤 역할을 하는지 이해했다.
- Docker가 왜 필요한지 설명할 수 있다.


# 3. 전체 서비스 흐름 그리기

코드를 작성하기 전에 사용자가 서비스를 어떻게 사용할지 먼저 상상해야 합니다.

추천하는 전체 흐름은 다음과 같습니다.

1. 사용자가 웹 페이지를 연다.
2. 캔버스 영역에 마우스로 숫자를 그린다.
3. 앱이 캔버스 이미지를 가져온다.
4. 이미지를 모델 입력에 맞게 전처리한다.
5. 전처리된 이미지를 화면에 보여준다.
6. ONNX 모델로 숫자를 예측한다.
7. 가장 가능성이 높은 숫자와 각 숫자의 확률을 보여준다.
8. 사용자가 그린 이미지와 예측 결과를 저장소 영역에 남긴다.

이 흐름을 먼저 정해두면, 각 기능을 따로 만들더라도 최종 서비스가 하나의 이야기처럼 연결됩니다.

완료 기준은 다음과 같습니다.

- 사용자가 입력하는 지점과 결과를 확인하는 지점을 구분할 수 있다.
- 전처리 단계가 입력과 추론 사이에 있어야 한다는 것을 이해했다.
- 추론 결과는 단순 label 하나뿐 아니라 확률 분포로도 보여줄 계획을 세웠다.
- 이미지 저장소 기능이 어떤 시점에 동작해야 하는지 정했다.


# 4. 작업 폴더 구조 설계하기

이번 미션은 파일이 여러 종류로 나뉘기 때문에 폴더 구조를 먼저 정해두는 것이 좋습니다.

추천 구조는 다음과 같습니다.

- `미션17/4팀_김도민/requirements.txt`: 필요한 라이브러리 목록
- `미션17/4팀_김도민/Dockerfile`: Docker 이미지 빌드 설정 파일
- `미션17/4팀_김도민/README.md`: 프로젝트 설명과 실행 방법

- 앱 코드, 모델 파일, 보고서 자료를 분리해서 관리할 계획을 세웠다.
- Docker에서 필요한 파일들이 어느 위치에 있어야 할지 대략적으로 이해했다.


# 5. Streamlit 화면 구성 계획 세우기

미션 안내에서는 서비스가 네 가지 주요 화면 영역을 가져야 한다고 설명합니다.

첫 번째는 입력 캔버스입니다. 사용자가 마우스로 숫자를 그리는 영역입니다. 이 영역은 `streamlit-drawable-canvas` 라이브러리를 활용하는 것이 미션 요구사항에 잘 맞습니다.

두 번째는 전처리 이미지 표시 영역입니다. 사용자가 그린 원본 이미지를 모델 입력에 맞게 바꾼 결과를 보여줍니다. 이 영역이 있으면 전처리가 제대로 되었는지 눈으로 확인할 수 있습니다.

세 번째는 모델 추론 결과 영역입니다. 0부터 9까지 각 숫자에 대한 예측 확률을 막대 차트 같은 방식으로 보여줍니다. 최종 예측 숫자만 보여주는 것보다, 모델이 얼마나 확신하는지 함께 보여주는 것이 좋습니다.

네 번째는 이미지 저장소 영역입니다. 사용자가 그린 이미지를 저장하고, 각 이미지의 예측 label과 확률을 함께 보여줍니다.

완료 기준은 다음과 같습니다.

- 화면을 입력, 전처리, 추론 결과, 저장소 영역으로 나눌 계획을 세웠다.
- 각 영역이 왜 필요한지 설명할 수 있다.
- 결과를 단순 텍스트뿐 아니라 시각적으로 보여줄 방법을 정했다.


# 6. MNIST ONNX 모델 준비 전략 정하기

이번 미션에서는 Github의 ONNX 모델 저장소에서 제공하는 MNIST ONNX 모델을 활용합니다.

모델 준비 방식은 두 가지 중 하나로 정할 수 있습니다.

첫 번째 방식은 모델 파일을 미리 다운로드해서 `models` 폴더에 넣어두는 방식입니다. 이 방식은 앱 실행 시 네트워크 연결에 덜 의존하므로 안정적입니다.

두 번째 방식은 앱이 실행될 때 모델 파일이 없으면 다운로드하도록 만드는 방식입니다. 이 방식은 제출 파일 크기를 줄일 수 있지만, 다운로드 URL 관리와 네트워크 오류 처리가 필요합니다.

기본 미션에서는 안정성을 우선해서, 모델을 `models` 폴더에 준비해두고 앱에서 해당 파일을 불러오는 방식을 추천합니다. 다만 보고서에는 모델 출처와 사용한 모델 파일명을 적어두는 것이 좋습니다.

완료 기준은 다음과 같습니다.

- 사용할 MNIST ONNX 모델의 위치를 정했다.
- 모델 파일을 직접 포함할지, 실행 중 다운로드할지 결정했다.
- 보고서에 모델 출처를 적어야 한다는 점을 기억했다.


# 7. 모델 로드와 캐싱 전략 정하기

Streamlit 앱은 사용자가 버튼을 누르거나 화면이 갱신될 때 코드가 다시 실행될 수 있습니다. 이때 모델을 매번 새로 불러오면 앱이 느려질 수 있습니다.

그래서 미션 안내에서는 세션 간 효율적인 모델 관리를 위해 캐싱 기능을 구현하라고 요구합니다.

캐싱의 목적은 `한 번 불러온 ONNX 모델 세션을 재사용하는 것`입니다. 모델 로드 함수와 추론 함수의 역할을 분리하면 구조가 깔끔해집니다.

모델 로드 단계에서 확인해야 할 것은 다음과 같습니다.

- ONNX Runtime으로 모델 파일을 불러올 수 있는가
- 모델의 입력 이름과 출력 이름을 확인할 수 있는가
- 같은 모델을 매번 새로 로드하지 않도록 캐싱할 수 있는가
- 모델 파일이 없을 때 사용자에게 이해 가능한 안내를 줄 수 있는가

완료 기준은 다음과 같습니다.

- 모델 로드와 이미지 추론을 별도 역할로 나누기로 했다.
- 캐싱이 필요한 이유를 설명할 수 있다.
- 모델 입력 이름과 출력 이름을 확인해야 한다는 점을 알고 있다.


# 8. 캔버스 입력 이미지 이해하기

사용자가 캔버스에 숫자를 그리면 앱은 그 결과를 이미지 데이터로 받습니다. 이 데이터는 보통 RGBA 형태처럼 색상 채널이 여러 개 포함된 이미지일 수 있습니다.

MNIST 모델은 일반적으로 28x28 크기의 흑백 숫자 이미지를 입력으로 기대합니다. 따라서 캔버스에서 받은 이미지를 그대로 모델에 넣으면 입력 형태가 맞지 않거나 예측 결과가 이상할 수 있습니다.

이 단계에서는 캔버스 입력이 어떤 형태로 들어오는지 확인하는 것이 중요합니다.

확인할 내용은 다음과 같습니다.

- 캔버스 이미지가 비어 있을 때 어떻게 처리할 것인가
- 배경색과 숫자 색이 모델이 학습한 MNIST 형태와 비슷한가
- 이미지에 알파 채널이 포함되어 있는가
- 숫자가 너무 작거나 한쪽으로 치우쳐 있을 때 전처리로 보정할 필요가 있는가

완료 기준은 다음과 같습니다.

- 캔버스 입력 이미지를 바로 모델에 넣으면 안 되는 이유를 이해했다.
- 빈 캔버스 상태를 처리해야 한다는 점을 기억했다.
- 캔버스 이미지와 MNIST 입력 이미지의 차이를 설명할 수 있다.


# 9. 이미지 전처리 기준 정하기

전처리는 이번 미션에서 가장 중요한 부분 중 하나입니다. 모델 추론 결과가 이상하다면 대부분 전처리 문제일 가능성이 큽니다.

전처리의 목표는 사용자가 그린 이미지를 MNIST 모델 입력 형식에 맞추는 것입니다.

일반적으로 확인해야 할 전처리 항목은 다음과 같습니다.

- 이미지를 흑백으로 변환한다.
- 모델 입력 크기인 28x28로 크기를 조정한다.
- 배경과 숫자의 밝기 방향을 MNIST와 맞춘다.
- 픽셀 값을 모델이 기대하는 범위로 정규화한다.
- 모델 입력 shape에 맞게 차원을 정리한다.

특히 색상 반전 여부가 중요합니다. MNIST는 보통 검은 배경에 밝은 숫자 형태로 학습되는 경우가 많습니다. 사용자가 흰 배경에 검은 숫자를 그리면 반전 처리가 필요할 수 있습니다.

완료 기준은 다음과 같습니다.

- 전처리 결과를 화면에 보여줄 계획을 세웠다.
- 28x28 흑백 이미지로 변환해야 한다는 점을 이해했다.
- 색상 반전, 정규화, 입력 shape가 추론 결과에 영향을 준다는 점을 알고 있다.


# 10. ONNX 추론 결과 해석하기

ONNX 모델의 출력은 보통 숫자 하나가 아니라 각 class에 대한 점수 또는 확률 형태로 나옵니다.

MNIST 숫자 분류라면 class는 0부터 9까지 총 10개입니다. 모델 출력에서 가장 큰 값을 가진 index가 예측 숫자가 됩니다.

이 단계에서는 추론 결과를 사용자에게 이해하기 쉬운 형태로 바꾸는 방법을 정해야 합니다.

확인할 내용은 다음과 같습니다.

- 모델 출력이 확률인지, 확률로 변환해야 하는 점수인지 확인한다.
- 가장 큰 값을 가진 숫자를 최종 예측 label로 정한다.
- 각 숫자별 값을 표나 차트로 보여줄 수 있게 정리한다.
- 최종 예측 확률을 함께 표시한다.

완료 기준은 다음과 같습니다.

- 모델 출력에서 최종 예측 숫자를 뽑는 방식을 이해했다.
- 0부터 9까지의 결과를 사용자에게 보여줄 계획을 세웠다.
- 확률 차트가 왜 필요한지 설명할 수 있다.


# 11. 예측 결과 시각화 방식 정하기

미션 안내에서는 각 레이블에 대한 예측 확률을 막대 차트 등으로 시각화하라고 요구합니다.

추천하는 결과 표시는 세 가지입니다.

첫째, 가장 크게 예측된 숫자를 눈에 잘 보이게 표시합니다. 예를 들어 `예측 결과: 7`처럼 최종 label을 명확히 보여줍니다.

둘째, 최종 예측 확률을 함께 보여줍니다. 사용자는 모델이 얼마나 확신하는지 확인할 수 있습니다.

셋째, 0부터 9까지 전체 확률을 막대 차트로 표시합니다. 이 차트가 있으면 7과 1처럼 헷갈릴 수 있는 숫자들 사이에서 모델이 어떻게 판단했는지 볼 수 있습니다.

보고서에는 이 화면을 캡처해서 넣으면 좋습니다. 특히 입력 캔버스, 전처리 이미지, 예측 차트가 함께 보이는 캡처가 있으면 프로젝트 설명이 쉬워집니다.

완료 기준은 다음과 같습니다.

- 최종 예측 label과 확률을 표시할 계획을 세웠다.
- 전체 class별 확률을 차트로 보여줄 계획을 세웠다.
- 보고서에 넣을 결과 화면 캡처를 어떤 장면으로 찍을지 정했다.


# 12. 이미지 저장소 기능 설계하기

미션 안내에는 사용자가 그린 이미지를 저장하고, 각 이미지와 함께 예측된 label 및 확률을 표시하는 이미지 저장소 기능이 포함되어 있습니다.

이 기능은 단순히 이미지를 파일로 저장하는 것보다, 사용자가 이전 입력 결과를 확인할 수 있게 만드는 데 의미가 있습니다.

추천 흐름은 다음과 같습니다.

1. 사용자가 숫자를 그리고 추론을 실행한다.
2. 앱이 원본 또는 전처리 이미지를 저장한다.
3. 저장할 때 예측 label과 확률도 함께 기록한다.
4. 화면 아래쪽 또는 별도 영역에 저장된 이미지 목록을 보여준다.

처음 구현할 때는 복잡한 데이터베이스를 쓰지 않아도 됩니다. 기본 미션에서는 세션 상태나 간단한 파일 저장 방식으로 충분합니다.

완료 기준은 다음과 같습니다.

- 저장할 이미지가 원본인지 전처리 이미지인지 정했다.
- 예측 label과 확률을 이미지와 함께 보여줄 계획을 세웠다.
- 저장소 기능을 너무 복잡하게 만들지 않기로 했다.


# 13. 로컬 실행과 기능 검증 계획 세우기

Docker로 가기 전에 먼저 로컬 환경에서 Streamlit 앱이 제대로 동작하는지 확인해야 합니다.

검증할 기능은 다음과 같습니다.

- 앱이 실행되는가
- 캔버스가 화면에 보이는가
- 사용자가 숫자를 그릴 수 있는가
- 빈 캔버스 상태에서 오류가 나지 않는가
- 전처리 이미지가 화면에 표시되는가
- ONNX 모델 추론이 성공하는가
- 예측 label과 확률 차트가 표시되는가
- 이미지 저장소에 결과가 남는가

이 단계에서 문제가 생기면 바로 보고서의 디버깅 기록 후보가 됩니다. 오류 메시지와 해결 과정을 짧게 기록해두면 마지막 보고서 작성이 쉬워집니다.

완료 기준은 다음과 같습니다.

- 로컬에서 서비스 전체 흐름을 한 번 이상 실행할 계획을 세웠다.
- 기능별 검증 항목을 알고 있다.
- 오류가 생겼을 때 기록할 준비를 했다.


# 14. Docker 컨테이너화 계획 세우기

이번 미션은 Dockerfile을 작성하여 서비스를 Docker 이미지로 빌드해야 합니다.

Docker의 목표는 `내 컴퓨터 환경에만 의존하지 않고 앱을 실행할 수 있게 만드는 것`입니다.

Dockerfile에는 일반적으로 다음 내용이 필요합니다.

- Python 실행 환경 선택
- 앱 파일과 필요한 파일 복사
- 필요한 라이브러리 설치
- Streamlit 앱 실행 설정
- 외부에서 접속할 포트 설정

Docker 단계에서 특히 확인해야 할 것은 모델 파일과 앱 파일이 컨테이너 안에서도 같은 경로로 접근 가능한지입니다. 로컬에서는 잘 되던 앱이 Docker 안에서 실패한다면, 파일 경로 문제일 가능성이 큽니다.

완료 기준은 다음과 같습니다.

- Dockerfile이 왜 필요한지 이해했다.
- 컨테이너 안에서도 모델 파일 경로가 맞아야 한다는 점을 기억했다.
- Docker 빌드 후 로컬에서 컨테이너 실행 테스트를 해야 한다는 점을 알고 있다.


# 15. Docker Hub 배포 계획 세우기

미션 제출 안내에는 Docker Hub URL을 보고서에 포함하라고 되어 있습니다. 따라서 Docker 이미지를 로컬에서 빌드하는 것만으로 끝나지 않고, Docker Hub에 업로드해야 합니다.

배포 흐름은 다음과 같습니다.

1. Docker Hub 계정을 준비한다.
2. 프로젝트용 이미지 이름을 정한다.
3. 로컬에서 Docker 이미지를 빌드한다.
4. 빌드한 이미지를 로컬에서 실행해본다.
5. 문제가 없으면 Docker Hub에 업로드한다.
6. 업로드된 Docker Hub URL을 보고서에 적는다.

보고서에는 Docker Hub URL만 적는 것이 아니라, 이 이미지를 통해 Streamlit 서비스를 실행할 수 있도록 컨테이너화했다고 설명하면 좋습니다.

완료 기준은 다음과 같습니다.

- Docker Hub URL이 최종 보고서에 필요하다는 점을 확인했다.
- 이미지 이름과 태그를 미리 정할 계획을 세웠다.
- 업로드 전에 로컬 Docker 실행 검증을 해야 한다는 점을 이해했다.

# 18. 심화 미션 진행 여부와 아이디어 정하기

심화 미션은 채점에 포함되지 않으므로 기본 미션을 먼저 완료한 뒤 여유가 있을 때 진행하는 것이 좋습니다.

심화 미션의 주제는 자유롭습니다. 지금까지 수업에서 다룬 CV 또는 NLP 모델을 활용해 원하는 형태의 서비스를 만들 수 있습니다.

진행한다면 기본 미션과 구분되도록 파일명에 `심화`를 붙여야 합니다.

가능한 심화 아이디어는 다음과 같습니다.

- 이미지 분류 모델을 활용한 간단한 사진 분류 서비스
- 감성 분석 모델을 활용한 문장 감정 분석 서비스
- 문서 요약 모델을 활용한 텍스트 요약 서비스
- 객체 탐지 또는 세그멘테이션 모델을 활용한 이미지 분석 서비스

다만 심화 미션 때문에 기본 미션 제출물이 흔들리면 안 됩니다. 기본 미션의 Streamlit 앱, Docker 배포, 보고서 PDF를 먼저 끝낸 뒤 진행 여부를 결정합니다.

완료 기준은 다음과 같습니다.

- 심화 미션은 선택 사항이라는 점을 이해했다.
- 기본 미션 완료 후 진행 여부를 결정하기로 했다.
- 진행한다면 기본 미션과 파일명을 명확히 구분할 계획을 세웠다.
