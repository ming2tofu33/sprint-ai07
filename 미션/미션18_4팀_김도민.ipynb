어느덧 스프린트의 마지막 미션입니다.
첫번째 미션을 기억해보세요. 우리는 파이썬 기초 문제를 풀었는데요.
어느덧 스스로 프론트엔드와 백엔드, 모델 서빙까지 전부 다룰 수 있는 AI 엔지니어가 되었습니다.
스스로와 여러분들의 동료들께 수고하셨다는 한마디 어떨까요!

## 프론트엔드 (Streamlit)
기능
- 영화 목록 표시

- 영화 추가
    - 입력: 제목, 개봉일, 감독, 장르, 포스터 URL
(심화)

- 리뷰 등록
    - 저장된 영화 선택
    - 작성자 이름, 리뷰 내용 입력
- 리뷰 감성 분석
    - 리뷰 작성 후 자동 실행
    - 감성 분석 결과 표시
- 리뷰 표시
    - 최근 10개 리뷰 표시
    - 항목: 영화 ID, 등록일, 리뷰 내용, 감성 분석 결과
배포
- Streamlit Cloud
참고
- 모든 데이터는 백엔드에서 관리
- Streamlit 내부에서 별도 저장 기능 사용 안 함

## 백엔드 (FastAPI)
기능
- 영화 관리
    - 등록: 제목, 개봉일, 감독, 장르, 포스터 URL (나무위키 참고)
    - 전체/특정 영화 조회
    - 특정 영화 삭제
(심화)
- 리뷰 관리
    - 등록, 전체/특정 영화 리뷰 조회, 삭제
- 평점 조회
    - 리뷰 감성 분석 점수의 평균
- 리뷰 감성 분석
    - 모델: 적절한 모델 리서치하여 적용

# 1. 최종 제출물 이해하기

이번 미션의 최종 제출물은 크게 `보고서 PDF`와 `코드 폴더`입니다. 코드 폴더는 반드시 `frontend`와 `backend`로 나누어 정리해야 합니다.

보고서 PDF에는 서비스 개요, 서비스 구조도, 데이터베이스 구조도인 ERD, FastAPI Docs 전체 캡처, 서비스 동작 캡처 이미지가 들어가야 합니다. 단순히 완성 화면만 보여주는 것이 아니라, Streamlit과 FastAPI가 어떻게 연결되고 데이터가 어디에 저장되는지 설명해야 합니다.

서비스 동작 캡처에는 영화 3개 이상 등록된 화면과 각 영화당 리뷰 10개 이상 등록된 화면이 필요합니다. 따라서 마지막에 급하게 데이터를 넣기보다, 처음부터 테스트용 영화와 리뷰 데이터를 준비해두는 것이 좋습니다.

완료 기준은 다음과 같습니다.

- 보고서 PDF에 들어갈 항목을 알고 있다.
- 코드 제출은 `frontend`, `backend` 폴더로 나누어야 한다는 점을 확인했다.
- 영화 3개 이상, 각 영화당 리뷰 10개 이상이 캡처에 필요하다는 점을 기억했다.
- FastAPI Docs 전체 캡처가 필수라는 점을 알고 있다.


# 2. 이번 미션에서 공부할 핵심 개념 정리하기

이번 미션은 AI 모델을 학습시키는 과제라기보다, AI 기능이 포함된 웹 서비스를 전체 구조로 만들어보는 과제입니다.

첫째, 프론트엔드와 백엔드 분리입니다. Streamlit은 사용자가 보는 화면을 담당하고, FastAPI는 영화와 리뷰 데이터를 관리하는 API 서버 역할을 합니다. Streamlit 내부에 별도 저장 기능을 두지 않고, 모든 데이터는 백엔드를 통해 관리해야 합니다.

둘째, API 기반 서비스 흐름입니다. 사용자가 Streamlit에서 영화 추가 버튼을 누르면 Streamlit이 FastAPI에 요청을 보내고, FastAPI가 데이터베이스에 저장한 뒤 결과를 다시 돌려주는 구조를 이해해야 합니다.

셋째, 데이터베이스 설계입니다. 영화 하나에 리뷰 여러 개가 연결되는 구조를 만들고, 이 관계를 ERD로 설명할 수 있어야 합니다.

넷째, 감성 분석 모델 적용입니다. 사용자가 리뷰를 작성하면 리뷰 텍스트를 모델에 넣어 긍정/부정 또는 점수 형태의 감성 분석 결과를 만들어야 합니다.

다섯째, 배포와 문서화입니다. 완성한 서비스를 실행 가능한 형태로 정리하고, FastAPI Docs와 서비스 화면을 캡처해서 보고서에 넣어야 합니다.

완료 기준은 다음과 같습니다.

- Streamlit과 FastAPI의 역할 차이를 설명할 수 있다.
- API 요청과 응답 흐름을 이해했다.
- 영화와 리뷰 사이의 관계를 설명할 수 있다.
- 감성 분석이 리뷰 등록 흐름 안에서 어디에 들어가는지 이해했다.


# 3. 기본 기능과 심화 기능 범위 정하기

미션 설명에는 일부 기능이 `(심화)`로 표시되어 있습니다. 하지만 제출 안내에는 각 영화당 리뷰 10개 이상 등록된 서비스 동작 캡처가 필요하다고 되어 있습니다. 따라서 안전하게 제출하려면 리뷰 등록, 리뷰 표시, 감성 분석 기능까지 구현 범위에 포함하는 것이 좋습니다.

기본적으로 반드시 가져가야 할 기능은 영화 목록 표시, 영화 추가, 영화 조회, 영화 삭제입니다. 여기에 제출 캡처 조건을 만족하기 위해 리뷰 등록, 리뷰 조회, 리뷰 감성 분석, 영화별 평균 감성 점수 표시까지 함께 계획합니다.

다만 처음부터 모든 기능을 한꺼번에 만들려고 하면 복잡해집니다. 먼저 영화 CRUD를 완성하고, 그 다음 리뷰 CRUD를 붙이고, 마지막으로 감성 분석 모델을 연결하는 순서가 좋습니다.

완료 기준은 다음과 같습니다.

- 리뷰와 감성 분석 기능을 제출 안정성을 위해 구현 대상으로 잡았다.
- 영화 기능을 먼저 만들고 리뷰 기능을 나중에 붙이는 순서로 진행하기로 했다.
- 심화 기능 때문에 기본 서비스 구조가 흔들리지 않도록 범위를 정했다.


# 4. 전체 서비스 구조 그리기

이번 서비스는 크게 네 부분으로 나누어 생각하면 됩니다.

첫 번째는 Streamlit 프론트엔드입니다. 사용자는 여기에서 영화 목록을 보고, 영화를 추가하고, 리뷰를 작성하고, 감성 분석 결과를 확인합니다.

두 번째는 FastAPI 백엔드입니다. Streamlit에서 들어온 요청을 받아 영화와 리뷰 데이터를 처리하고, 필요한 결과를 응답으로 돌려줍니다.

세 번째는 데이터베이스입니다. 영화 정보와 리뷰 정보가 저장됩니다. Streamlit이 직접 저장하지 않고 FastAPI가 데이터베이스와 통신하는 구조를 유지해야 합니다.

네 번째는 감성 분석 모델입니다. 리뷰 텍스트가 들어오면 모델이 감성 분석 결과를 만들고, 그 결과를 리뷰 데이터와 함께 저장합니다.

추천 흐름은 다음과 같습니다.

1. 사용자가 Streamlit 화면을 연다.
2. Streamlit이 FastAPI에 영화 목록을 요청한다.
3. FastAPI가 데이터베이스에서 영화 목록을 가져와 응답한다.
4. 사용자가 영화를 추가하거나 리뷰를 작성한다.
5. Streamlit이 FastAPI에 등록 요청을 보낸다.
6. FastAPI가 데이터를 검증하고 데이터베이스에 저장한다.
7. 리뷰 등록 시 감성 분석 모델이 실행된다.
8. Streamlit이 갱신된 영화, 리뷰, 평균 점수를 다시 보여준다.

완료 기준은 다음과 같습니다.

- 사용자가 보는 화면과 데이터가 저장되는 위치를 구분할 수 있다.
- Streamlit이 FastAPI를 통해서만 데이터를 다루는 구조를 이해했다.
- 감성 분석 모델이 리뷰 등록 흐름에 연결된다는 점을 파악했다.


# 5. 작업 폴더 구조 설계하기

이번 미션은 프론트엔드와 백엔드를 분리해서 제출해야 하므로 폴더 구조를 먼저 정하는 것이 중요합니다.

추천 구조는 다음과 같습니다.

- `미션18/4팀_김도민/frontend`: Streamlit 앱 파일과 프론트엔드 실행에 필요한 파일
- `미션18/4팀_김도민/backend`: FastAPI 서버 파일, 데이터베이스 관련 파일, 모델 관련 파일
- `미션18/4팀_김도민/report`: 보고서 PDF와 캡처 이미지 정리
- `미션18/4팀_김도민/README.md`: 전체 실행 방법과 프로젝트 설명

프론트엔드와 백엔드를 나누는 이유는 역할이 다르기 때문입니다. Streamlit은 화면과 사용자 입력을 담당하고, FastAPI는 데이터 처리와 API 응답을 담당합니다.

또한 보고서에 넣을 캡처 이미지는 처음부터 `report` 폴더에 모아두면 마지막에 PDF를 만들 때 편합니다.

완료 기준은 다음과 같습니다.

- 제출 폴더 이름을 미션 안내 형식에 맞게 정했다.
- `frontend`와 `backend`를 분리해서 관리할 계획을 세웠다.
- 보고서 자료를 별도 폴더에 모을 계획을 세웠다.


# 6. 데이터베이스 구조와 ERD 설계하기

이번 미션에서 데이터베이스는 영화 정보와 리뷰 정보를 저장하는 역할을 합니다. 복잡한 구조보다 요구사항을 정확히 만족하는 단순한 구조가 좋습니다.

가장 중요한 테이블은 `movies`와 `reviews`입니다.

`movies`에는 영화 제목, 개봉일, 감독, 장르, 포스터 URL 같은 정보가 들어갑니다. 미션 설명에서 포스터 URL은 나무위키 등을 참고할 수 있다고 되어 있으므로, 실제 이미지가 화면에 표시될 수 있는 URL을 준비해야 합니다.

`reviews`에는 어떤 영화에 달린 리뷰인지, 작성자 이름, 리뷰 내용, 등록일, 감성 분석 결과, 감성 점수 같은 정보가 들어갑니다.

관계는 영화 1개에 리뷰 여러 개가 연결되는 1:N 구조입니다. 보고서의 ERD에서는 이 관계가 가장 잘 보이도록 표현하면 됩니다.

완료 기준은 다음과 같습니다.

- 영화 테이블에 들어갈 필드를 정했다.
- 리뷰 테이블에 들어갈 필드를 정했다.
- 영화와 리뷰가 1:N 관계라는 점을 ERD에 표시할 계획을 세웠다.
- 감성 분석 결과와 점수를 리뷰 데이터에 함께 저장할지 결정했다.


# 7. FastAPI 백엔드 API 설계하기

FastAPI는 이번 서비스의 데이터 처리 중심입니다. 프론트엔드에서 필요한 기능을 API로 제공해야 합니다.

영화 관리 API에서는 영화 등록, 전체 영화 조회, 특정 영화 조회, 특정 영화 삭제가 필요합니다. 영화 등록 시에는 제목, 개봉일, 감독, 장르, 포스터 URL을 받아 저장합니다.

리뷰 관리 API에서는 리뷰 등록, 전체 리뷰 조회, 특정 영화의 리뷰 조회, 리뷰 삭제가 필요합니다. 리뷰 등록 시에는 저장된 영화 중 하나를 선택하고, 작성자 이름과 리뷰 내용을 입력받습니다.

감성 분석 관련 API는 리뷰 등록 과정에 포함시키는 것이 자연스럽습니다. 사용자가 리뷰를 등록하면 백엔드에서 자동으로 감성 분석을 실행하고, 그 결과를 리뷰와 함께 저장합니다.

평점 조회 API는 특정 영화의 리뷰 감성 점수를 평균 내어 반환하는 방식으로 설계할 수 있습니다.

완료 기준은 다음과 같습니다.

- 영화 관리 API 목록을 정했다.
- 리뷰 관리 API 목록을 정했다.
- 리뷰 등록 시 감성 분석이 자동 실행되도록 흐름을 정했다.
- FastAPI Docs에서 API 설명이 확인 가능해야 한다는 점을 기억했다.


# 8. 감성 분석 모델 적용 전략 정하기

감성 분석 모델은 리뷰 내용을 입력받아 긍정/부정 또는 점수 형태의 결과를 만드는 역할입니다.

가장 먼저 결정할 것은 어떤 언어의 리뷰를 대상으로 할지입니다. 한국어 리뷰를 사용할 계획이라면 한국어 감성 분석이 가능한 모델을 찾는 것이 좋습니다. 영어 리뷰를 사용할 계획이라면 영어 감성 분석 모델을 선택해도 됩니다.

두 번째로 고려할 것은 모델의 무게입니다. 이번 미션은 백엔드, 프론트엔드, 데이터베이스까지 함께 다뤄야 하므로 너무 무거운 모델을 쓰면 실행과 배포가 어려워질 수 있습니다. 따라서 실행 속도와 설치 난이도까지 고려해서 모델을 선택해야 합니다.

세 번째로 정할 것은 저장 형식입니다. 예를 들어 감성 결과는 `긍정`, `부정`, `중립`처럼 label로 저장하고, 감성 점수는 평균 계산이 가능하도록 숫자로 저장할 수 있습니다.

보고서에는 사용한 모델, 모델 선택 이유, 경량화나 실행 속도 측면에서 고려한 점을 적으면 좋습니다.

완료 기준은 다음과 같습니다.

- 한국어 리뷰를 쓸지 영어 리뷰를 쓸지 정했다.
- 감성 분석 모델 선택 기준을 정했다.
- 감성 결과를 label과 score 중 어떤 형태로 저장할지 결정했다.
- 보고서에 모델 선택 이유를 설명할 계획을 세웠다.


# 9. Streamlit 프론트엔드 화면 설계하기

Streamlit은 사용자가 직접 보는 화면입니다. 화면은 기능별로 구역을 나누면 이해하기 쉽습니다.

첫 번째 영역은 영화 목록입니다. 등록된 영화의 제목, 포스터 이미지, 감독, 장르, 개봉일, 평균 감성 점수 또는 평균 평점을 보여줍니다.

두 번째 영역은 영화 추가 폼입니다. 제목, 개봉일, 감독, 장르, 포스터 URL을 입력받아 백엔드에 등록 요청을 보냅니다.

세 번째 영역은 영화 선택 영역입니다. 저장된 영화 중 하나를 선택하면 해당 영화의 리뷰 목록과 평균 점수를 확인할 수 있게 합니다.

네 번째 영역은 리뷰 등록 폼입니다. 작성자 이름과 리뷰 내용을 입력하면 백엔드에 리뷰 등록 요청을 보내고, 감성 분석 결과를 받아 표시합니다.

다섯 번째 영역은 리뷰 표시 영역입니다. 최근 리뷰 또는 선택한 영화의 리뷰를 보여주고, 리뷰 내용과 감성 분석 결과를 함께 표시합니다.

완료 기준은 다음과 같습니다.

- 화면을 영화 목록, 영화 추가, 영화 선택, 리뷰 등록, 리뷰 표시 영역으로 나눴다.
- 각 영역이 어떤 API를 호출해야 하는지 대략적으로 연결했다.
- 평균 점수나 감성 결과가 사용자에게 보이도록 계획했다.


# 10. 영화 등록과 조회 흐름 정하기

영화 등록은 이번 서비스의 기본 기능입니다. 사용자는 제목, 개봉일, 감독, 장르, 포스터 URL을 입력하고 등록합니다.

등록이 성공하면 영화 목록에 바로 반영되어야 합니다. 사용자는 등록된 영화의 포스터와 기본 정보를 확인할 수 있어야 합니다.

제출 캡처에는 영화 3개 이상이 필요하므로, 미리 등록할 영화 후보를 정해두면 좋습니다. 포스터 URL은 화면에 잘 표시되는지 확인해야 합니다. 포스터가 깨지면 서비스 완성도가 떨어져 보일 수 있습니다.

영화 삭제 기능도 백엔드 요구사항에 포함되어 있습니다. 삭제할 때는 해당 영화에 연결된 리뷰를 어떻게 처리할지도 정해야 합니다. 기본적으로는 영화 삭제 시 연결된 리뷰도 함께 삭제되도록 설계하는 것이 관리하기 쉽습니다.

완료 기준은 다음과 같습니다.

- 등록할 영화 정보 항목을 확인했다.
- 제출 캡처용 영화 3개 이상을 준비할 계획을 세웠다.
- 영화 등록 후 목록이 갱신되어야 한다는 흐름을 정했다.
- 영화 삭제 시 연결된 리뷰 처리 방식을 고민했다.


# 11. 리뷰 등록과 감성 분석 흐름 정하기

리뷰 기능은 제출 캡처 조건과 연결되므로 중요합니다. 각 영화당 리뷰 10개 이상을 등록해야 하므로, 테스트용 리뷰 문장을 미리 준비해두는 것이 좋습니다.

추천 흐름은 다음과 같습니다.

1. 사용자가 Streamlit에서 저장된 영화를 선택한다.
2. 작성자 이름과 리뷰 내용을 입력한다.
3. Streamlit이 FastAPI에 리뷰 등록 요청을 보낸다.
4. FastAPI가 리뷰 내용을 감성 분석 모델에 전달한다.
5. 모델이 감성 결과와 점수를 만든다.
6. FastAPI가 리뷰, 감성 결과, 감성 점수를 데이터베이스에 저장한다.
7. Streamlit이 최신 리뷰 목록을 다시 불러와 화면에 표시한다.

이 흐름에서 중요한 점은 감성 분석이 사용자가 따로 실행하는 기능이 아니라 리뷰 등록 후 자동으로 실행되는 기능이라는 점입니다.

완료 기준은 다음과 같습니다.

- 리뷰 등록 시 필요한 입력값을 정했다.
- 각 영화당 리뷰 10개 이상을 준비할 계획을 세웠다.
- 리뷰 등록과 동시에 감성 분석이 실행되는 흐름을 정했다.
- 리뷰 목록에 감성 결과가 함께 표시되도록 계획했다.


# 12. 평균 평점 또는 감성 점수 계산 방식 정하기

미션 설명에는 영화 목록에서 평균 평점 표시가 옵션으로 들어가 있고, 백엔드 심화 항목에는 리뷰 감성 분석 점수의 평균 조회가 포함되어 있습니다.

리뷰에 별도의 별점 입력을 받지 않는다면, 감성 분석 점수를 평균 내어 영화별 평균 점수처럼 보여줄 수 있습니다. 예를 들어 긍정 리뷰는 높은 점수, 부정 리뷰는 낮은 점수로 저장하면 평균 계산이 가능합니다.

중요한 것은 계산 기준을 보고서에서 설명할 수 있어야 한다는 점입니다. 단순히 숫자를 보여주는 것보다, 이 숫자가 감성 분석 결과를 바탕으로 계산되었다고 설명하면 좋습니다.

화면에서는 영화 목록 카드나 표에 평균 감성 점수를 함께 보여주면 사용자가 영화별 리뷰 분위기를 빠르게 비교할 수 있습니다.

완료 기준은 다음과 같습니다.

- 평균 점수를 감성 분석 점수 기반으로 계산할지 정했다.
- 감성 label과 숫자 score의 관계를 설명할 수 있게 정리했다.
- 영화 목록에서 평균 점수를 보여줄 위치를 정했다.


# 13. 로컬 실행과 통신 검증 계획 세우기

배포 전에 먼저 로컬에서 백엔드와 프론트엔드가 제대로 통신하는지 확인해야 합니다.

추천 검증 순서는 백엔드를 먼저 실행하고 FastAPI Docs에서 API가 정상 동작하는지 확인하는 것입니다. 그 다음 Streamlit을 실행해서 프론트엔드 화면에서 같은 기능이 동작하는지 확인합니다.

검증할 내용은 다음과 같습니다.

- FastAPI 서버가 실행되는가
- FastAPI Docs에서 영화 등록과 조회가 되는가
- FastAPI Docs에서 리뷰 등록과 조회가 되는가
- 리뷰 등록 시 감성 분석 결과가 생성되는가
- Streamlit 화면에서 영화 목록이 보이는가
- Streamlit에서 영화 추가가 되는가
- Streamlit에서 리뷰 등록 후 결과가 갱신되는가
- 프론트엔드가 직접 저장하지 않고 백엔드 API를 통해 데이터를 가져오는가

오류가 생기면 보고서의 문제 해결 과정에 넣을 수 있으므로, 오류 메시지와 해결 방법을 간단히 기록해두는 것이 좋습니다.

완료 기준은 다음과 같습니다.

- 백엔드 단독 검증 후 프론트엔드 통신 검증을 진행하기로 했다.
- FastAPI Docs를 기능 테스트와 캡처에 모두 활용할 계획을 세웠다.
- 오류와 해결 과정을 보고서 자료로 남기기로 했다.


# 14. 배포 계획 세우기

미션 안내에는 프론트엔드 배포로 Streamlit Cloud가 제시되어 있습니다. 따라서 Streamlit 앱이 배포 환경에서도 백엔드 API에 접근할 수 있도록 구조를 생각해야 합니다.

로컬에서는 Streamlit과 FastAPI가 같은 컴퓨터에서 실행되지만, 배포 후에는 Streamlit Cloud에서 실행되는 프론트엔드가 외부에서 접근 가능한 백엔드 주소를 사용해야 할 수 있습니다.

따라서 API 주소를 코드 안에 고정하기보다 설정값으로 관리하는 것이 좋습니다. 로컬 실행 주소와 배포용 주소를 구분할 수 있어야 배포 시 수정이 편합니다.

백엔드 배포 방식은 수업에서 안내한 방식이 있다면 그 방식을 우선합니다. 별도 안내가 없다면, 최소한 로컬에서 FastAPI가 정상 동작하고 Docs 캡처가 가능해야 하며, Streamlit Cloud에서 백엔드에 접근 가능한 방법을 고민해야 합니다.

완료 기준은 다음과 같습니다.

- Streamlit Cloud 배포가 필요한 점을 확인했다.
- 배포 환경에서 백엔드 API 주소가 중요하다는 점을 이해했다.
- 로컬 주소와 배포 주소를 구분해서 관리할 계획을 세웠다.
- 백엔드 배포 방식은 수업 안내를 우선 확인하기로 했다.


# 15. 필수 캡처 자료 준비하기

보고서에는 캡처 이미지가 많이 필요합니다. 마지막에 한꺼번에 캡처하려고 하면 빠지는 화면이 생기기 쉬우므로, 필요한 캡처 목록을 미리 정리해두는 것이 좋습니다.

필수 캡처는 다음과 같습니다.

- FastAPI Docs 전체 화면
- 영화 등록 API 설명과 요청/응답 예시가 보이는 화면
- 리뷰 등록 API 설명과 요청/응답 예시가 보이는 화면
- Streamlit 영화 목록 화면
- 영화 3개 이상이 등록된 화면
- 특정 영화를 선택한 화면
- 각 영화당 리뷰 10개 이상이 등록된 화면
- 리뷰별 감성 분석 결과가 표시된 화면
- 평균 감성 점수 또는 평균 평점이 표시된 화면

캡처를 찍을 때는 보고서에 설명하기 쉬운 장면을 선택하는 것이 좋습니다. 예를 들어 영화 목록, 리뷰 등록 폼, 리뷰 결과가 한 화면에 잘 드러나면 서비스 흐름을 설명하기 쉽습니다.

완료 기준은 다음과 같습니다.

- 보고서에 필요한 캡처 목록을 정했다.
- 영화 3개와 각 영화당 리뷰 10개 이상을 캡처 전에 준비하기로 했다.
- FastAPI Docs 캡처가 누락되지 않도록 체크리스트에 넣었다.


# 16. 보고서 PDF 작성 계획

보고서는 길게 쓰는 것보다 미션 요구사항이 빠짐없이 보이게 구성하는 것이 중요합니다.

추천 구성은 다음과 같습니다.

## 1. 서비스 개요

영화 정보와 사용자 리뷰를 관리하고, 리뷰 감성 분석 결과를 표시하는 Streamlit과 FastAPI 기반 웹 애플리케이션이라고 설명합니다.

## 2. 서비스 구조도

Streamlit 프론트엔드, FastAPI 백엔드, 데이터베이스, 감성 분석 모델이 어떻게 연결되는지 그림으로 보여줍니다.

## 3. 데이터베이스 구조도 ERD

영화 테이블과 리뷰 테이블의 필드, 1:N 관계를 설명합니다.

## 4. 주요 기능 설명

영화 등록/조회/삭제, 리뷰 등록/조회/삭제, 감성 분석, 평균 점수 표시를 기능별로 설명합니다.

## 5. 감성 분석 모델 설명

# 17. 제출 폴더와 ZIP 구성 정리하기

제출 전에는 폴더 구조와 파일 구성이 요구사항에 맞는지 확인해야 합니다.

추천 제출 구성은 다음과 같습니다.

- `frontend`: Streamlit 앱, 프론트엔드 requirements, 프론트엔드 실행 설명
- `backend`: FastAPI 앱, 데이터베이스 관련 파일, 감성 분석 모델 관련 파일, 백엔드 requirements, 백엔드 실행 설명
- `report`: 보고서 PDF, 보고서에 사용한 캡처 이미지
- `README.md`: 전체 프로젝트 개요, 로컬 실행 방법, 배포 관련 안내

ZIP으로 묶기 전에 불필요한 캐시 파일, 가상환경 폴더, 큰 임시 파일이 들어가지 않도록 확인해야 합니다. 제출자는 코드를 받아 실행할 수 있어야 하므로 실행 방법이 README에 정리되어 있으면 좋습니다.

완료 기준은 다음과 같습니다.

- `frontend`와 `backend` 폴더가 분리되어 있다.
- 보고서 PDF가 제출 폴더 안에 포함되어 있다.
- 실행에 필요한 의존성 파일과 설명이 포함되어 있다.
- 불필요한 임시 파일을 제출 ZIP에서 제외할 계획을 세웠다.
