# 1. 최종 제출물 이해하기

이 미션은 모델 성능을 최고로 끌어올리는 과제라기보다, 학습된 모델을 여러 저장 형식으로 변환하고 다시 사용할 수 있는지 검증하는 과제입니다.

기본 미션의 최종 제출물은 크게 세 가지입니다.

- `modeling.ipynb`: 모델 학습, 기본 `.pth` 저장, 양자화 `.pth` 저장, `.onnx` 변환을 수행하는 노트북
- `inference.ipynb`: 저장된 모델을 다시 불러와 추론이 되는지 확인하는 노트북
- 요약 보고서 PDF: 모델 파일 이름, 파일 용량 비교 캡처, 디버깅과 오류 해결 과정을 정리한 문서

제출 폴더에는 최소한 아래 모델 파일 3개가 만들어져야 합니다.

- `mission_16_mnist_mlp.pth`: 기본 PyTorch 모델
- `mission_16_mnist_mlp_quantized.pth`: 양자화된 PyTorch 모델
- `mission_16_mnist_mlp.onnx`: ONNX 형식 모델

파일 이름은 보고서 캡처와 제출물에서 한 번에 확인될 수 있도록 미션 요구 형식에 맞춰 통일하는 것이 좋습니다.


# 2. 양자화 핵심 개념 정리하기

양자화는 모델 안의 숫자를 더 가벼운 형태로 바꿔서 모델 용량을 줄이는 작업입니다.

일반적인 PyTorch 모델의 가중치는 보통 `float32` 형태입니다. `float32`는 숫자 하나를 표현하는 데 32비트를 사용합니다. 양자화는 이 값을 더 작은 정수 표현, 예를 들어 8비트 정수에 가까운 형태로 바꿔 저장하거나 계산하도록 만드는 과정입니다.

이번 미션에서 기억할 핵심은 세 가지입니다.

- 장점: 모델 파일 용량이 줄어들 수 있습니다.
- 장점: CPU 환경에서 추론 속도가 빨라질 수 있습니다.
- 단점: 숫자를 더 거칠게 표현하기 때문에 정확도가 조금 떨어질 수 있습니다.

보고서에서는 양자화를 어렵게 설명하기보다, `기본 모델 대비 양자화 모델의 파일 용량이 얼마나 줄었고, 추론 결과나 정확도는 얼마나 유지되었는지`를 중심으로 설명하면 됩니다.


# 3. 모델 선택 전략 정하기

추천 모델은 MNIST 숫자 분류용 MLP입니다.

MNIST는 0부터 9까지 손글씨 숫자 이미지를 분류하는 공개 벤치마크 데이터셋입니다. 이미지 크기는 28x28이고, 입력을 펼치면 784개의 숫자 값이 됩니다.

MLP를 추천하는 이유는 다음과 같습니다.

- 모델 구조가 단순해서 학습과 디버깅이 쉽습니다.
- `Linear` 계층 중심이라 PyTorch 동적 양자화를 적용하기 쉽습니다.
- ONNX 변환이 비교적 안정적입니다.
- 정확도보다 저장 형식 변환이 핵심인 이번 미션에 잘 맞습니다.

CNN도 가능하지만, 처음부터 CNN을 선택하면 양자화와 ONNX 변환에서 확인할 점이 늘어납니다. 이번 과제에서는 `복잡한 모델을 고르는 것`보다 `세 가지 포맷으로 저장하고 추론 검증을 끝내는 것`이 더 중요합니다.

# 5. `modeling.ipynb` 전체 흐름 설계하기

`modeling.ipynb`는 모델을 만들고 저장 형식을 변환하는 핵심 노트북입니다.

이 노트북에 들어갈 흐름은 다음과 같습니다.

1. 필요한 라이브러리와 실행 환경을 확인합니다.
2. MNIST 데이터를 불러오고 학습용, 평가용 데이터로 나눕니다.
3. 간단한 MLP 모델 구조를 정의합니다.
4. 모델을 학습합니다.
5. 평가 데이터로 기본 모델의 정확도를 확인합니다.
6. 기본 모델을 `.pth` 파일로 저장합니다.
7. 기본 모델을 양자화한 뒤, 양자화 모델을 `.pth` 파일로 저장합니다.
8. 기본 모델을 ONNX 형식으로 변환하여 `.onnx` 파일로 저장합니다.
9. 저장된 세 모델 파일의 용량을 비교합니다.

중요한 점은 이 노트북에서 모든 결과 파일이 실제로 생성되어야 한다는 것입니다. 단순히 학습 정확도만 출력하고 끝나면 미션 요구사항을 충족하지 못합니다.


# 6. MNIST 데이터와 평가 기준 정하기

이번 과제에서 사용할 평가 데이터셋은 MNIST 테스트 데이터입니다.

MNIST는 각 이미지에 정답 label이 있으므로, 모델이 예측한 숫자와 실제 숫자가 같은지 비교해서 정확도를 계산할 수 있습니다.

확인해야 할 항목은 다음과 같습니다.

- 입력 이미지 크기가 28x28인지
- 모델 입력 형태가 학습과 추론에서 일관되는지
- label이 0부터 9까지의 정수인지
- 평가용 데이터에서 기본 모델과 양자화 모델의 정확도를 비교할 수 있는지

보고서에는 정확도를 너무 깊게 분석하지 않아도 됩니다. 다만 `기본 모델이 정상적으로 숫자를 분류했고, 양자화 모델도 비슷한 수준으로 동작했다`는 정도는 수치로 보여주는 것이 좋습니다.


# 7. 기본 PyTorch 모델 저장 전략 정하기

기본 `.pth` 파일은 PyTorch에서 모델을 다시 불러오기 위한 저장 파일입니다.

저장할 때는 모델의 구조와 가중치를 다시 복원할 수 있어야 합니다. 보통은 모델 구조를 코드에 정의해두고, 학습된 가중치를 `.pth` 파일에 저장합니다.

이번 미션에서 기본 모델 저장 단계의 목표는 다음과 같습니다.

- 학습된 MLP 모델을 저장합니다.
- 저장된 파일 이름을 `mission_16_mnist_mlp.pth`처럼 명확하게 정합니다.
- 저장 직후 파일이 실제로 존재하는지 확인합니다.
- 가능하면 저장된 모델을 다시 불러와 같은 입력에 대해 예측이 되는지 확인합니다.

이 단계가 안정적으로 끝나야 양자화와 ONNX 변환도 비교 기준을 가질 수 있습니다.


# 8. 양자화 모델 저장 전략 정하기

양자화 모델은 기본 모델보다 가볍게 만든 버전입니다.

이번 미션에서는 MLP의 `Linear` 계층을 중심으로 동적 양자화를 적용하는 방향이 가장 무난합니다. 동적 양자화는 학습을 다시 길게 하지 않고도 적용할 수 있어서 과제 진행 부담이 작습니다.

양자화 단계에서 확인할 내용은 다음과 같습니다.

- 기본 모델을 기준으로 양자화 모델을 만듭니다.
- 양자화 모델을 `mission_16_mnist_mlp_quantized.pth` 이름으로 저장합니다.
- 저장된 파일 용량이 기본 `.pth`보다 줄었는지 확인합니다.
- 양자화 모델로도 평가 데이터 추론이 되는지 확인합니다.
- 기본 모델 대비 정확도가 크게 무너지지 않았는지 확인합니다.

보고서에서는 `양자화 후 파일 용량이 감소했고, 정확도는 큰 차이 없이 유지되었다` 또는 `용량은 줄었지만 정확도 변화가 있었다`처럼 결과를 솔직하게 정리하면 됩니다.


# 9. ONNX 변환 전략 정하기

ONNX는 PyTorch가 아닌 다른 실행 환경에서도 모델을 사용할 수 있도록 만든 공통 모델 형식입니다.

기본 미션에서는 학습된 PyTorch 모델을 ONNX 파일로 변환하고, ONNX Runtime으로 다시 추론해보는 것이 목표입니다.

ONNX 변환 단계에서 확인할 내용은 다음과 같습니다.

- 모델을 평가 모드로 바꾼 뒤 변환합니다.
- MNIST 입력과 같은 형태의 더미 입력을 준비합니다.
- ONNX 파일 이름을 `mission_16_mnist_mlp.onnx`로 저장합니다.
- 저장된 ONNX 파일이 실제로 존재하는지 확인합니다.
- ONNX Runtime으로 불러와 추론 결과가 나오는지 확인합니다.

ONNX 변환에서 가장 자주 생기는 문제는 입력 shape이 맞지 않는 것입니다. 따라서 학습, 저장, 추론에서 입력 형태를 일관되게 맞추는 것이 중요합니다.


# 10. 파일 용량 비교와 캡처 준비하기

요약 보고서에는 추출된 모델들과 용량을 한 번에 확인할 수 있는 이미지 캡처가 필요합니다.

따라서 모델 파일 3개가 저장된 뒤, 같은 폴더에서 파일 이름과 용량이 보이도록 정리해야 합니다.

캡처에 포함되면 좋은 정보는 다음과 같습니다.

- `mission_16_mnist_mlp.pth` 파일 이름과 용량
- `mission_16_mnist_mlp_quantized.pth` 파일 이름과 용량
- `mission_16_mnist_mlp.onnx` 파일 이름과 용량
- 세 파일이 같은 `models` 폴더 안에 있다는 점

보고서에서는 단순히 캡처만 넣기보다, `양자화 모델은 기본 PyTorch 모델보다 용량이 줄었고, ONNX 모델은 다른 실행 환경에서 사용할 수 있는 교환 형식으로 저장했다`고 짧게 설명하면 좋습니다.


# 11. `inference.ipynb` 추론 검증 계획 세우기

`inference.ipynb`는 저장된 모델이 실제로 다시 사용 가능한지 검증하는 노트북입니다.

이 노트북에서는 학습을 다시 하면 안 됩니다. 이미 저장된 모델 파일을 불러와 추론만 수행해야 합니다.

추천 흐름은 다음과 같습니다.

1. `models` 폴더에 세 모델 파일이 있는지 확인합니다.
2. MNIST 테스트 이미지 몇 장을 준비합니다.
3. 기본 `.pth` 모델을 불러와 예측 label을 확인합니다.
4. 양자화 `.pth` 모델을 불러와 예측 label을 확인합니다.
5. `.onnx` 모델을 ONNX Runtime으로 불러와 예측 label을 확인합니다.
6. 같은 이미지에 대해 세 모델의 예측 결과가 비슷한지 비교합니다.
7. 가능하면 테스트 데이터 전체 또는 일부에 대해 정확도를 계산합니다.

이 단계가 끝나면 `저장된 모델 파일이 실제 추론에 사용할 수 있다`는 것을 증명할 수 있습니다.


# 12. 디버깅 기록 방식 정하기

보고서에는 미션 중 디버깅, 오류 사항, 오류 해결 과정이 포함되어야 합니다.

오류가 생겼을 때는 그냥 해결하고 넘어가지 말고, 아래 형식으로 짧게 기록해두면 보고서 작성이 쉬워집니다.

| 항목 | 기록할 내용 |
| --- | --- |
| 오류 상황 | 어떤 단계에서 문제가 생겼는지 |
| 오류 메시지 | 핵심 오류 문장 |
| 원인 추정 | 왜 문제가 생겼다고 판단했는지 |
| 해결 방법 | 어떤 수정으로 해결했는지 |
| 확인 결과 | 수정 후 어떤 결과가 나왔는지 |

이번 미션에서 자주 나올 수 있는 오류는 다음과 같습니다.

오류가 없었다면, 시도한 검증 과정을 정리해도 됩니다. 예를 들어 `저장 직후 재로드 테스트를 수행하여 파일 손상 여부를 확인했다`처럼 작성할 수 있습니다.

# 14. 심화 미션 진행 전략 정하기
